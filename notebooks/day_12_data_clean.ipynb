```python
# Cell 1 — Import library
import pandas as pd
```

```python
# Cell 2 — Load dataset
df = pd.read_csv("../data/raw/customer_support_tickets.csv")
```

```python
# Cell 3 — Count missing values in each column
df.isnull().sum()
```

```python
# Cell 4 — Count duplicate rows
df.duplicated().sum()
```

```python
# Cell 5 — INvestigate rows with missing resolution.
df[df["Resolution"].isnull()][
    ["Ticket Status", "Resolution", "First Response Time",
     "Time to Resolution", "Customer Satisfaction Rating"]
].head(10)
```

```python
# Cell 6 — Convert date of purchase to datetime
df["Date of Purchase"] = pd.to_datetime(
    df["Date of Purchase"],
    errors="coerce"
)
```

```python
# Cell 7 — Convert first response time to datetime
df["First Response Time"] = pd.to_datetime(
    df["First Response Time"],
    errors="coerce"
)
```

```python
# Cell 8 — Convert Time to Resolution to datetime
df["Time to Resolution"] = pd.to_datetime(
    df["Time to Resolution"],
    errors="coerce"
)
```

```python
# Cell 9 — Fill the blank cells under Resolution with the value "Not Resolved"
df["Resolution"] = df["Resolution"].fillna("Not Resolved")
```

```python
# Cell 10 — Check data types
df.dtypes
```

```python
# Cell 11 — Validate cleaning
df.info()
df.isnull().sum()
```
