```python
# Cell 1 — Import library
import pandas as pd
```

```python
# Cell 2 — Load dataset
df = pd.read_csv("../data/raw/customer_support_tickets.csv")

df.head()
```

```python
# Cell 3 - Inspect dataset shape and structure
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

df.info()
```

```python
# Cell 4 — Count missing values in each column
missing_values = df.isnull().sum()

missing_values
```

- ### Missing Values

Finding:
- The following columns contained missing values:
    - Resolution
    - First Response Time
    - Time to resolution
    - Customer Satisfaction Rating

Interpretation:
- The resolution is missing when the issue related to that ticket was not resolved and therefore a Time for the resolution could not be   filled. The first response time contains missing values when no response was attached to that ticket. The missing values on CUstomer Satisfaction rating occur when the customer does not provide feedback on the customer satisfaction page.

Decision:
- Missing Resolution values will be represented as "Not Resolved".
- Missing timestamps will remain missing if they correspond to tickets that have not reached that stage.

Reason:
- A ticket whose issue is not resolved should be tagged as not resolved under the resolution column to facilitate addessing those tickets who still need to be resolved.
- Missing operational timestamps can contain useful business meaning. A ticket that has not been resolved should not be given an artificial resolution time.

```python
# Cell 5 — Count duplicate rows
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
# Cell 6 - Remove Duplicate Rows
df = df.drop_duplicates()

print("Rows after removing duplicates:", len(df))
```

### Duplicate Rows

Finding:
- No duplicates were found

Decision:
- Exact duplicate records were removed.

Reason:
- Exact duplicate rows would cause tickets to be counted more than once and could distort later analysis.

```python
# Cell 7 — Inspect column names
df.columns.tolist()

for column in df.columns:
    print(repr(column))
```

```python
# Cell 8 - Verify Whitespaces
df.columns = df.columns.str.strip()

df.columns.tolist()
```

### Column Names

Finding:
- Column names were inspected for leading and trailing whitespace.

Decision:
- Leading and trailing whitespace was removed from column names.

Reason:
- Consistent column names reduce errors when referencing fields in Python.

```python
# Cell 9 — Inspect data types
df.dtypes
```

```python
# Cell 10 — Convert Date of Purchase to date time
df["Date of Purchase"] = pd.to_datetime(
    df["Date of Purchase"],
    errors="coerce"
)
```

```python
# Cell 11 — Convert First Response Time to date time
df["First Response Time"] = pd.to_datetime(df["First Response Time"], errors="coerce")
```

```python
# Cell 12 — Convert Time to Resolution to date time
df["Time to Resolution"] = pd.to_datetime(df["Time to Resolution"], errors="coerce")

# Inspect
df[
    [
        "Date of Purchase",
        "First Response Time",
        "Time to Resolution"
    ]
].dtypes
```

### Date and Time Columns

Finding:
- The date/time columns were initially stored in a format that needed explicit conversion.

Decision:
- Date-related columns were converted with pd.to_datetime().
- Invalid values are converted to NaT using errors="coerce".

Reason:
- Datetime values are required for reliable time-based analysis later in the project.

```python
# Cell 13 — Investigate rows with missing resolution.
unresolved_tickets = df.loc[
    df["Resolution"].isnull(),
    [
        "Ticket Status",
        "Resolution",
        "First Response Time",
        "Time to Resolution",
        "Customer Satisfaction Rating"
    ]
]

unresolved_tickets.head(10)
```

```python
# Cell 14 - Count tickets with missing resolution
print(
    "Tickets with missing Resolution:",
    df["Resolution"].isnull().sum()
)
```

```python
# Cell 15 — Fill the blank cells under Resolution with the value "Not Resolved"
df["Resolution"] = df["Resolution"].fillna("Not Resolved")

# Verify
df["Resolution"].isnull().sum()
```

### Missing Resolution Values

Finding:
- Some tickets did not contain resolution text.

Interpretation:
- These records represent tickets without a recorded resolution.

Decision:
- Missing Resolution values were replaced with "Not Resolved".

Reason:
- This keeps unresolved tickets identifiable instead of treating them as unexplained missing text.

```python
# Cell 16 - Inspect whitespaces in all text columns
text_columns = df.select_dtypes(include="object").columns

text_columns
```

```python
# Cell 17 - Remove leading/trailing whitespace from text columns
for column in text_columns:
    df[column] = df[column].str.strip()
```

```python
# Cell 18 — Inspect important categorical values
print("Ticket Types:")
print(df["Ticket Type"].value_counts(dropna=False))

print("\nTicket Statuses:")
print(df["Ticket Status"].value_counts(dropna=False))
```

```python
# Cell 19 — Normalize categorical capitalization
df["Ticket Type"] = df["Ticket Type"].str.strip().str.title()
df["Ticket Status"] = df["Ticket Status"].str.strip().str.title()

# Verify

print(df["Ticket Type"].value_counts(dropna=False))
print()
print(df["Ticket Status"].value_counts(dropna=False))
```

### Categorical Consistency

Finding:
- Ticket Type and Ticket Status were inspected for inconsistent spacing and capitalization.

Decision:
- Leading/trailing spaces were removed and category capitalization was standardized.

Reason:
- Values referring to the same category should not be counted separately because of formatting differences.
