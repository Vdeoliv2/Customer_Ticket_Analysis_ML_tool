## Purpose of the Split

### Make the model identify and classify the new ticket on a particular type just by reading its description.

**Input (X):** Ticket Description

**Output (Y):** Ticket Type

**Model Type:** Classification

**Business Use:** The support team can use the predicted ticket type to automatically categorize incoming customer tickets and route them to the appropriate support team or workflow.

```python
# Cell 1 — Load cleaned data
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv("../data/processed/customer_support_tickets_clean.csv")
print(f"Duplicated tickets: {df.duplicated().sum()}")
invalid_type = df["Ticket Type"].fillna("").str.strip().eq("").sum()
print(f"Number of invalid Tickets due to missing Ticket Type: {invalid_type}")
invalid_description = df["Ticket Description"].fillna("").str.strip().eq("").sum()
print(f"Number of invalid Tickets due to missing Ticket Description: {invalid_description}")
```

```python
# Cell 2 — Define X and Y
X = df["Ticket Description"]
y = df["Ticket Type"]
```

```python
# Cell 3 — Create 80/20 stratified split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

```python
# Cell 4 — Verify split sizes
print(f"There are {len(X_train)} descriptions in X_train")
print(f"There are {len(y_train)} labels in y_train")
print(f"There are {len(X_test)} descriptions in X_test")
print(f"There are {len(y_test)} labels in y_test")
```

```python
# Cell 5 — Verify X/Y alignment
split_sizes_valid = (
    len(X_train) == len(y_train)
    and len(X_test) == len(y_test)
    and len(X_train) + len(X_test) == len(X)
    and len(y_train) + len(y_test) == len(y)
)
if split_sizes_valid:
    print("The train/test split sizes are valid.")
else:
    print("The train/test split sizes are NOT valid.")
```

```python
# Cell 6 — Compare class distributions
comparison = pd.DataFrame({"Full": y.value_counts(dropna=False, normalize=True) * 100,
                            "Train": y_train.value_counts(dropna=False, normalize=True) * 100,
                            "Test": y_test.value_counts(dropna=False, normalize=True) * 100
                            })
display(comparison.style.format("{:.2f}%"))
```

```python
# Cell 7 — Inspect several examples
train_examples = pd.DataFrame({
    "Ticket Description": X_train,
    "Ticket Type": y_train
})

train_examples.sample(10, random_state=42)
```

## Day 23 Conclusions

- The dataset was divided into training and testing sets using an 80/20 split. The training set will be used to develop the ticket classification model, while the test set will remain separate so that the model can later be evaluated on unseen ticket descriptions.

- `Ticket Description` was used as the input feature (X), while `Ticket Type` was used as the target variable (y).

- Stratification was performed using `Ticket Type` so that the proportions of the five ticket categories remained approximately the same in the full dataset, training set, and test set.

- The training input and target contain the same number of observations, and the testing input and target also contain the same number of observations. This confirms that the train/test split produced corresponding X and y sets.

- The class-distribution comparison shows that the training and testing sets maintain approximately the same Ticket Type percentages as the complete dataset.

- A fixed `random_state` was used so that the same train/test split can be reproduced when the notebook is run again.

- The resulting training and testing sets are ready for the next stage, where the ticket descriptions will be converted into numerical features before training a classification model.
