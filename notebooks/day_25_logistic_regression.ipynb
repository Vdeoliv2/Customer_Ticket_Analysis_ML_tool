```python
# Cell 1 — Load cleaned data
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

df = pd.read_csv("../data/processed/customer_support_tickets_clean.csv")
print(f"Duplicated tickets: {df.duplicated().sum()}")
invalid_type = df["Ticket Type"].fillna("").str.strip().eq("").sum()
print(f"Invalid or empty Ticket Type values: {invalid_type}")
invalid_description = df["Ticket Description"].fillna("").str.strip().eq("").sum()
print(f"Invalid or empty Ticket Description values: {invalid_description}")
```

```python
# Cell 2 — Define X and Y
X = df["Ticket Description"]
y = df["Ticket Type"]
```

```python
# Cell 3 — Create 80/20 stratified split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state= 42
)
```

```python
# Cell 4 — Create Vectorizer
vectorizer = TfidfVectorizer()
```

```python
# Cell 5 — Fit and transform training data
X_train_tfidf = vectorizer.fit_transform(X_train)
```

```python
# Cell 6 — Transform test data
X_test_tfidf = vectorizer.transform(X_test)
```

```python
# Cell 7 — Create Logistic Regression Model
model = LogisticRegression()
```

```python
# Cell 8 — Train Logistic Regression Model
model.fit(X_train_tfidf, y_train)
```

```python
# Cell 9 — Generate Test Predictions
y_pred = model.predict(X_test_tfidf)
```

```python
# Cell 10 — Inspect Sample Predictions
comparison = pd.DataFrame({
    "Description": X_test.reset_index(drop=True),
    "Actual": y_test.reset_index(drop=True),
    "Predicted": y_pred
})

comparison.sample(10, random_state=42)
```
