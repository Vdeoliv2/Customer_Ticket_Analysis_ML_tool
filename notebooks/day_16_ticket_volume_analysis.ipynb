```python
# Cell 1 — Import library
import pandas as pd
```

```python
# Cell 2 — Load dataset
df = pd.read_csv("../data/processed/customer_support_tickets_clean.csv")
```

```python
# Cell 3 —  Verify suspicious variations
print(sorted(df["Ticket Type"].dropna().unique()))
df["Ticket Type"] = df["Ticket Type"].str.strip()
print("Original unique values:",
      df["Ticket Type"].nunique())

print("Case-insensitive unique values:",
      df["Ticket Type"].str.lower().nunique())
```

```python
# Cell 4 — ANALYSIS QUESTION 1. Which ticket types occur most frequently?
 
invalid = df["Ticket Type"].fillna("").str.strip().eq("").sum()
freq = df["Ticket Type"].value_counts(dropna=False)
percentage = df["Ticket Type"].value_counts(dropna=False, normalize=True) * 100

 
# 1. Tickets per ticket type
print()
print("Ticket Count by Type")
print(freq)
print()

# 2. Percentage per ticket type
print("Percentage by type:")
print(percentage.map(lambda x: f"{x:.2f}%"))
print()

print(f"The most common ticket type is {freq.idxmax()} with {freq.max()} tickets.")
print(f"The least common ticket type is {freq.idxmin()} with {freq.min()} tickets.")
print(f"The difference of ticket percentage between the most and the leas common ticket types is {(percentage.max() - percentage.min()):.2f}%.")
```

## Business Observations

### Ticket Volume Analysis

**Most common ticket type:** Refund request — 1,752 tickets (20.7%)

**Least common ticket type:** Billing inquiry — 1,634 tickets (19.3%)

### Key observations

- Refund requests are the most common ticket type, with 1,752 tickets
  (20.69% of all tickets).

- Billing inquiries are the least common ticket type, with 1,634 tickets
  (19.29% of all tickets).

- Ticket volume is distributed relatively evenly across all five categories.
  The difference between the largest category (20.69%) and smallest category
  (19.29%) is only 1.39 percentage points.

- Refund requests and technical issues have nearly identical volumes:
  1,752 and 1,747 tickets respectively, a difference of only 5 tickets.

### Business interpretation

No single ticket type overwhelmingly dominates customer-support demand.
This suggests that the support operation needs to handle a relatively balanced
mix of refund requests, technical issues, cancellation requests, product
inquiries, and billing inquiries.

Although refund requests have the highest volume, ticket volume alone is not
enough to conclude that refunds are the most problematic area. The next
analyses should compare categories by customer satisfaction, resolution time,
priority, and unresolved-ticket volume to determine which areas actually
require the most attention.
